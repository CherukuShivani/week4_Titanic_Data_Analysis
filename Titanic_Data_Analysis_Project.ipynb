```python
titanic = train.copy()

print("Dataset copied successfully!")
```

```python
print("Missing Values:")
print(titanic.isnull().sum())
```

```python
missing_percentage = (
    titanic.isnull().sum() / len(titanic)
) * 100

print(missing_percentage)
```

```python
titanic['Age'] = titanic['Age'].fillna(
    titanic['Age'].median()
)

print("Missing Age values handled!")
```

```python
titanic['Embarked'] = titanic['Embarked'].fillna(
    titanic['Embarked'].mode()[0]
)

print("Missing Embarked values handled!")
```

```python
titanic['Cabin_Known'] = titanic['Cabin'].notnull().astype(int)

print("Cabin information processed!")
```

```python
titanic = titanic.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(titanic.shape)
```

```python
print("Remaining Missing Values:")
print(titanic.isnull().sum())
```

```python
plt.figure(figsize=(6,6))

survival_counts = titanic['Survived'].value_counts()

plt.pie(
    survival_counts,
    labels=['Not Survived', 'Survived'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Titanic Passenger Survival Percentage')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=titanic,
    x='Pclass',
    y='Survived'
)

plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=titanic,
    x='Age',
    bins=30,
    kde=True
)

plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=titanic,
    x='Fare',
    bins=30,
    kde=True
)

plt.title('Fare Distribution of Titanic Passengers')
plt.xlabel('Ticket Fare')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
titanic['FamilySize'] = (
    titanic['SibSp'] + titanic['Parch'] + 1
)

print(titanic[['SibSp', 'Parch', 'FamilySize']].head())
```

```python
plt.figure(figsize=(9,5))

sns.barplot(
    data=titanic,
    x='FamilySize',
    y='Survived'
)

plt.title('Survival Rate by Family Size')
plt.xlabel('Family Size')
plt.ylabel('Survival Rate')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=titanic,
    x='Pclass',
    y='Survived',
    hue='Sex'
)

plt.title('Survival Rate by Gender and Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')

plt.show()
```

```python
print("TITANIC DATA ANALYSIS SUMMARY")
print("-----------------------------")

print("Total Passengers:", len(titanic))

print("Total Survivors:",
      titanic['Survived'].sum())

print("Total Deaths:",
      (titanic['Survived'] == 0).sum())

print("Overall Survival Rate:",
      round(titanic['Survived'].mean() * 100, 2), "%")
```

```python
gender_results = titanic.groupby('Sex')['Survived'].mean() * 100

print("Survival Rate by Gender:")
print(gender_results.round(2))
```

```python
class_results = titanic.groupby('Pclass')['Survived'].mean() * 100

print("Survival Rate by Passenger Class:")
print(class_results.round(2))
```

```python
titanic['AgeGroup'] = pd.cut(
    titanic['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child', 'Teenager', 'Young Adult', 'Adult', 'Senior']
)

age_results = titanic.groupby(
    'AgeGroup', observed=False
)['Survived'].mean() * 100

print("Survival Rate by Age Group:")
print(age_results.round(2))
```

```python
family_results = titanic.groupby(
    'FamilySize'
)['Survived'].mean() * 100

print("Survival Rate by Family Size:")
print(family_results.round(2))
```

```python
print("TITANIC DATA ANALYSIS - KEY FINDINGS")
print("------------------------------------")

print("Total Passengers:", len(titanic))

print("Total Survivors:", titanic['Survived'].sum())

print("Total Deaths:", (titanic['Survived'] == 0).sum())

print("Overall Survival Rate:",
      round(titanic['Survived'].mean() * 100, 2), "%")

print("\nGender with highest survival rate:",
      gender_results.idxmax())

print("Passenger class with highest survival rate:",
      class_results.idxmax())
```
